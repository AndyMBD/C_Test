```python
size = 384
```

```python
import math
offset = math.pi/2 # pi/2 = 90 degrees
```

```python
#
# Generate SVM-D (0/50) scheme and table
#
import math
import matplotlib.pyplot as plt
import numpy as np
import os
import sys
ulist = []
vlist = []
wlist = []
for index in range (size):
    sine_u = math.sin((2 * math.pi * (index / size)) + offset)
    sine_v = math.sin((2 * math.pi * ((index / size) + 2/3)) + offset)
    sine_w = math.sin((2 * math.pi * ((index / size) + 1/3)) + offset)
    ph_u = max(((sine_u - min(sine_v, sine_w)) / 2) * (2 / math.sqrt(3)), 0)
    ph_v = max(((sine_v - min(sine_u, sine_w)) / 2) * (2 / math.sqrt(3)), 0)
    ph_w = max(((sine_w - min(sine_u, sine_v)) / 2) * (2 / math.sqrt(3)), 0)
    ulist.append((ph_u * 65535) + 0.5)
    vlist.append((ph_v * 65535) + 0.5)
    wlist.append((ph_w * 65535) + 0.5)
plt.plot(range(size), ulist, color="green")
plt.plot(range(size), vlist, color="blue")
plt.plot(range(size), wlist, color="red")
plt.xlabel("index")
plt.ylabel("Q0.16")
plt.yticks(np.arange(0, 65535, 0x2000))
plt.xticks(np.arange(0, size, size/6))
plt.ylim([0,65535])
plt.xlim([0,size])
plt.grid(True)
plt.title("SVM-D {} scheme".format(size))
plt.savefig(os.path.join(".", "lib_svmd_scheme_{}.png".format(size)))
plt.show()

table_size = int((size * 5) / 3)

ulist = []
vlist = []
wlist = []
for index in range (table_size):
    sine_u = math.sin((2 * math.pi * (index / size)) + offset)
    sine_v = math.sin((2 * math.pi * ((index / size) + 2/3)) + offset)
    sine_w = math.sin((2 * math.pi * ((index / size) + 1/3)) + offset)
    ph_u = max(((sine_u - min(sine_v, sine_w)) / 2) * (2 / math.sqrt(3)), 0)
    ph_v = max(((sine_v - min(sine_u, sine_w)) / 2) * (2 / math.sqrt(3)), 0)
    ph_w = max(((sine_w - min(sine_u, sine_v)) / 2) * (2 / math.sqrt(3)), 0)
    ulist.append((ph_u * 65535) + 0.5)
    vlist.append((ph_v * 65535) + 0.5)
    wlist.append((ph_w * 65535) + 0.5)
plt.plot(range(table_size), ulist, color="green")
plt.xlabel("index")
plt.ylabel("Q0.16")
plt.yticks(np.arange(0, 65535, 0x2000))
plt.xticks(np.arange(0, table_size, table_size/5))
plt.ylim([0,65535])
plt.xlim([0,table_size])
plt.grid(True)
plt.title("SVM-D {} table".format(size))
plt.savefig(os.path.join(".", "lib_svmd_table_{}.png".format(size)))
plt.show()

if debug is True:
    f = sys.stdout
else:
    f = open("lib_svmd_table_{}.h".format(size), "w")
print('#define MOTOR_SVM_D_PERIOD (' + str(size) + 'u)', file=f)
print('#define MOTOR_SVM_D_SIZE (' + str(table_size) + 'u)', file=f)
print('const uint16_t Motor_SVM_D[] = {', file=f)
for index in range(size):
    print('    0x{0:04X}u,'.format(int(ulist[index])), file=f)
for index in range(int(table_size - size)):
    print('    0x{0:04X}u,'.format(int(ulist[index])), file=f)    
print('};', file=f)
```

```python
#
# Generate SVM-LH (0/50) scheme and table
#
import math
import matplotlib.pyplot as plt
import numpy as np
import os
import sys
ulist = []
vlist = []
wlist = []
rlist = []
slist = []
tlist = []
for index in range (size):
    sine_u = math.sin((2 * math.pi * (index / size)) + offset)
    sine_v = math.sin((2 * math.pi * ((index / size) + 2/3)) + offset)
    sine_w = math.sin((2 * math.pi * ((index / size) + 1/3)) + offset)
    max_sine = max([sine_u, sine_v, sine_w])
    min_sine = min([sine_u, sine_v, sine_w])
    if -min_sine >= max_sine: # Lowest sine at reference 0
        ph_u = (sine_u - min_sine) / math.sqrt(3)
        ph_v = (sine_v - min_sine) / math.sqrt(3)
        ph_w = (sine_w - min_sine) / math.sqrt(3)
    else: # Highest sine at reference 1
        ph_u = 1 + (sine_u - max_sine) / math.sqrt(3)
        ph_v = 1 + (sine_v - max_sine) / math.sqrt(3)
        ph_w = 1 + (sine_w - max_sine) / math.sqrt(3)    
    if debug is True:
        rlist.append((sine_u * 32768) + 32767)
        slist.append((sine_v * 32768) + 32767)
        tlist.append((sine_w * 32768) + 32767)
    ulist.append((ph_u * 65535) + 0.5)
    vlist.append((ph_v * 65535) + 0.5)
    wlist.append((ph_w * 65535) + 0.5)
if debug is True:    
    plt.plot(range(size), rlist, color="lightgreen")
    plt.plot(range(size), slist, color="lightblue")
    plt.plot(range(size), tlist, color="lightcoral")
plt.plot(range(size), ulist, color="green")
plt.plot(range(size), vlist, color="blue")
plt.plot(range(size), wlist, color="red")
plt.xlabel("index")
plt.ylabel("Q0.16")
plt.yticks(np.arange(0, 65535, 0x2000))
plt.xticks(np.arange(0, size, size/6))
plt.ylim([0,65535])
plt.xlim([0,size])
plt.grid(True)
plt.title("SVM-LH {} scheme".format(size))
plt.savefig(os.path.join(".", "lib_svmlh_scheme_{}.png".format(size)))
plt.show()

table_size = int((size * 5) / 3)

ulist = []
vlist = []
wlist = []
for index in range (table_size):
    sine_u = math.sin((2 * math.pi * (index / size)) + offset)
    sine_v = math.sin((2 * math.pi * ((index / size) + 2/3)) + offset)
    sine_w = math.sin((2 * math.pi * ((index / size) + 1/3)) + offset)    
    max_sine = max([sine_u, sine_v, sine_w])
    min_sine = min([sine_u, sine_v, sine_w])
    if -min_sine >= max_sine: # Lowest sine at reference 0
        ph_u = (sine_u - min_sine) / math.sqrt(3)
        ph_v = (sine_v - min_sine) / math.sqrt(3)
        ph_w = (sine_w - min_sine) / math.sqrt(3)
    else: # Highest sine at reference 1
        ph_u = 1 + (sine_u - max_sine) / math.sqrt(3)
        ph_v = 1 + (sine_v - max_sine) / math.sqrt(3)
        ph_w = 1 + (sine_w - max_sine) / math.sqrt(3)    
    ulist.append((ph_u * 65535) + 0.5)
    vlist.append((ph_v * 65535) + 0.5)
    wlist.append((ph_w * 65535) + 0.5)
plt.plot(range(table_size), ulist, color="green")
plt.xlabel("index")
plt.ylabel("Q0.16")
plt.yticks(np.arange(0, 65535, 0x2000))
plt.xticks(np.arange(0, table_size, table_size/5))
plt.ylim([0,65535])
plt.xlim([0,table_size])
plt.grid(True)
plt.title("SVM-LH {} table".format(size))
plt.savefig(os.path.join(".", "lib_svmlh_table_{}.png".format(size)))
plt.show()

if debug is True:
    f = sys.stdout
else:
    f = open("lib_svmlh_table_{}.h".format(size), "w")
print('#define MOTOR_SVM_LH_PERIOD (' + str(size) + 'u)', file=f)
print('#define MOTOR_SVM_LH_SIZE (' + str(table_size) + 'u)', file=f)
print('const uint16_t Motor_SVM_LH[] = {', file=f)
for index in range(size):
    print('    0x{0:04X}u,'.format(int(ulist[index])), file=f)
for index in range(int(table_size - size)):
    print('    0x{0:04X}u,'.format(int(ulist[index])), file=f)    
print('};', file=f)
```

```python
#
# Generate SVM-M (25/75) scheme and table
#
import math
import matplotlib.pyplot as plt
import numpy as np
import os
import sys
ulist = []
vlist = []
wlist = []
sulist = []
svlist = []
swlist = []
slist = []
for index in range (size):
    sine_u = math.sin((2 * math.pi * (index / size)) + offset)
    sine_v = math.sin((2 * math.pi * ((index / size) + 2/3)) + offset)
    sine_w = math.sin((2 * math.pi * ((index / size) + 1/3)) + offset)
    min_uvw = min(sine_u, sine_v, sine_w) / 2
    max_uvw = max(sine_u, sine_v, sine_w) / 2    
    star = (0 - min_uvw - max_uvw)    
    ph_u = (sine_u + star) * (2 / math.sqrt(3))
    ph_v = (sine_v + star) * (2 / math.sqrt(3))
    ph_w = (sine_w + star) * (2 / math.sqrt(3))
    if debug is True:
        sulist.append((sine_u * 32767) + 0.5)
        svlist.append((sine_v * 32767) + 0.5)
        swlist.append((sine_w * 32767) + 0.5)
    ulist.append((ph_u * 32767) + 0.5)
    vlist.append((ph_v * 32767) + 0.5)
    wlist.append((ph_w * 32767) + 0.5)
    if debug is True:
        slist.append((star * 32767) + 0.5)
plt.plot(range(size), ulist, color="green")
plt.plot(range(size), vlist, color="blue")
plt.plot(range(size), wlist, color="red")
if debug is True:
    plt.plot(range(size), sulist, color="lightgreen")
    plt.plot(range(size), svlist, color="lightblue")
    plt.plot(range(size), swlist, color="lightcoral")
    plt.plot(range(size), slist, color="yellow")
plt.xlabel("index")
plt.ylabel("Q0.16")
plt.yticks(np.arange(-32768, 32767, 2*4096))
plt.xticks(np.arange(0, size, size/6))
plt.ylim([-32768,32767])
plt.xlim([0,size])
plt.grid(True)
plt.title("SVM-M {} scheme".format(size))
plt.savefig(os.path.join(".", "lib_svmm_scheme_{}.png".format(size)))
plt.show()

table_size = int((size * 5) / 3)

ulist = []
vlist = []
wlist = []
for index in range (table_size):
    sine_u = math.sin((2 * math.pi * (index / size)) + offset)
    sine_v = math.sin((2 * math.pi * ((index / size) + 2/3)) + offset)
    sine_w = math.sin((2 * math.pi * ((index / size) + 1/3)) + offset)
    min_uvw = min(sine_u, sine_v, sine_w) / 2
    max_uvw = max(sine_u, sine_v, sine_w) / 2
    star = (0 - min_uvw - max_uvw)    
    ph_u = (sine_u + star) * (2 / math.sqrt(3))
    ph_v = (sine_v + star) * (2 / math.sqrt(3))
    ph_w = (sine_w + star) * (2 / math.sqrt(3))
    ulist.append((ph_u * 32767) + 0.5)
    vlist.append((ph_v * 32767) + 0.5)
    wlist.append((ph_w * 32767) + 0.5)
plt.plot(range(table_size), ulist, color="green")
plt.xlabel("index")
plt.ylabel("Q0.16")
plt.yticks(np.arange(-32768, 32767, 2*4096))
plt.xticks(np.arange(0, table_size, table_size/5))
plt.ylim([-32768,32767])
plt.xlim([0,table_size])
plt.grid(True)
plt.title("SVM-M {} table".format(size))
plt.savefig(os.path.join(".", "lib_svmm_table_{}.png".format(size)))
plt.show()

if debug is True:
    f = sys.stdout
else:
    f = open("lib_svmm_table_{}.h".format(size), "w")
print('#define MOTOR_SVM_M_PERIOD (' + str(size) + 'u)', file=f)
print('#define MOTOR_SVM_M_SIZE (' + str(table_size) + 'u)', file=f)
print('const uint16_t Motor_SVM_M[] = {', file=f)
for index in range(size):
    print('    {0},'.format(int(ulist[index])), file=f)
for index in range(int(table_size - size)):
    print('    {0},'.format(int(ulist[index])), file=f)    
print('};', file=f)

print(max(ulist))
```
